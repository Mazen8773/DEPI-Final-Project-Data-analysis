# HR Data Analysis — Week 2
## Proposed Analysis Questions

### 1. Employee Age and Job Satisfaction
- **Question:** Is employee age associated with job satisfaction?
- **Columns:** Age, JobSatisfaction.
- **Analysis:** Compare average satisfaction across age groups.
- **Business value:** Help HR understand satisfaction patterns among different age groups.

### 2. Job Satisfaction by Department
- **Question:** Which departments have the lowest average job satisfaction?
- **Columns:** Department, JobSatisfaction.
- **Analysis:** Calculate average satisfaction for each department.
- **Business value:** Identify departments that may need further investigation or employee engagement initiatives.

### 3. Job Satisfaction and Employee Attrition
- **Question:** Does job satisfaction differ between employees who leave and employees who stay?
- **Columns:** Attrition, JobSatisfaction.
- **Analysis:** Compare satisfaction between attrition groups.
- **Business value:** Help HR investigate potential employee retention problems.

### 4. Salary and Employee Attrition
- **Question:** Is salary associated with employee attrition?
- **Columns:** Salary, Attrition.
- **Analysis:** Compare salary distributions between attrition groups.
- **Business value:** Help management evaluate whether compensation may be related to employee turnover.

### 5. Overtime and Employee Satisfaction
- **Question:** Is overtime associated with job satisfaction and employee attrition?
- **Columns:** Overtime, JobSatisfaction, Attrition.
- **Analysis:** Compare satisfaction and attrition across overtime groups.
- **Business value:** Help HR investigate possible workload and retention issues.

### 6. Training and Employee Performance
- **Question:** Is training participation associated with manager performance ratings?
- **Columns:** TrainingOpportunitiesTaken, ManagerRating.
- **Analysis:** Compare manager ratings across training participation levels.
- **Business value:** Help management evaluate training outcomes and identify opportunities for further investigation.

## Tools
- SQL for querying, grouping, and aggregating data.
- Python Pandas for data analysis.
- Matplotlib for data visualization.

A set of well-defined analysis questions that can guide the next phase of the HR data analysis project and support organizational decision-making.
