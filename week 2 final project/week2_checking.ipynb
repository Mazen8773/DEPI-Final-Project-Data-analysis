# 1. Load and inspect the data


```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("hr_dataset_cleaned.csv")

print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())
```

# Check employee and review records

```python
print("Total rows:", len(df))
print("Unique employees:", df["employeeid"].nunique())
print("Duplicate rows:", df.duplicated().sum())

if "performanceid" in df.columns:
    print("Unique performance records:", df["performanceid"].nunique())

if "attrition" in df.columns:
    print("\nAttrition distribution:")
    print(df["attrition"].value_counts())
```

# Check the number of reviews per employee

```python
print("Total records:", len(df))
print("Unique employees:", df["employeeid"].nunique())

print("\nRecords per employee:")
print(df["employeeid"].value_counts().value_counts().sort_index())
```

# Check the performance review identifiers

```python
print("All columns:")
print(df.columns.tolist())

print("\nPerformance IDs:")
if "performanceid" in df.columns:
    print("Unique performance IDs:", df["performanceid"].nunique())
    print("Duplicate performance IDs:", df["performanceid"].duplicated().sum())
```

# Check which columns change between records

```python
employee_columns = [
    "firstname",
    "lastname",
    "gender",
    "age",
    "department",
    "salary"
]

for col in employee_columns:
    if col in df.columns:
        print(f"{col}: {df.groupby('employeeid')[col].nunique().max()} maximum distinct values per employee")
```

#### checking whether repeated performance IDs belong to the same employee and whether those records are exact duplicates.

```python
# Check duplicated performance IDs
duplicate_reviews = df[
    df["performanceid"].duplicated(keep=False)
].sort_values("performanceid")

print("Rows with repeated performance IDs:", len(duplicate_reviews))

print("\nNumber of employees per performance ID:")
print(
    duplicate_reviews.groupby("performanceid")["employeeid"]
    .nunique()
    .value_counts()
)

print("\nExact duplicate rows:", df.duplicated().sum())

print("\nSample repeated performance IDs:")
print(
    duplicate_reviews[
        ["employeeid", "performanceid", "reviewdate",
         "jobsatisfaction", "managerrating"]
    ].head(20)
)
```
